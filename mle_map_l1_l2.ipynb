```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_breast_cancer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score,confusion_matrix 
```

```python
#load dataset
a=load_breast_cancer()
x=a.data
y=a.target

print("Shape: ",x.shape)
print("Classes: ",np.unique(y))
```

```python
#Split
x_train,x_test,y_train,y_test=train_test_split(
    x,y,random_state=42,test_size=0.2
)
```

```python
#Feature Scaling
s=StandardScaler()
x_train=s.fit_transform(x_train)
x_test=s.transform(x_test)
```

```python
m1=LogisticRegression(C=np.inf,max_iter=5000,random_state=42)
m1.fit(x_train,y_train)
p1=m1.predict(x_test)
```

```python
#MLE L2
m2=LogisticRegression(C=1.0,l1_ratio=0,solver="lbfgs",max_iter=5000,random_state=42)
m2.fit(x_train,y_train)
p2=m2.predict(x_test)
```

```python
#MLE L1
m3=LogisticRegression(C=1.0,l1_ratio=1,solver="liblinear",max_iter=5000,random_state=42)
m3.fit(x_train,y_train)
p3=m1.predict(x_test)
```

```python
#Evaluate MLE
print("\nMLE")
print("Accuracy:",accuracy_score(y_test,p1))
print("Precision:",precision_score(y_test,p1))
print("Recall:",recall_score(y_test,p1))
print("F1-Score:",f1_score(y_test,p1))
print("Confusion Matrix:\n",confusion_matrix(y_test,p1))
```

```python
#  Evaluate MAP L2
print("\nMAP L2")
print("Accuracy:", accuracy_score(y_test, p2))
print("Precision:", precision_score(y_test, p2))
print("Recall:", recall_score(y_test, p2))
print("F1 Score:", f1_score(y_test, p2))
print("Confusion Matrix:\n", confusion_matrix(y_test, p2))
```

```python
#  Evaluate MAP L1
print("\nMAP L1")
print("Accuracy:", accuracy_score(y_test, p3))
print("Precision:", precision_score(y_test, p3))
print("Recall:", recall_score(y_test, p3))
print("F1 Score:", f1_score(y_test, p3))
print("Confusion Matrix:\n", confusion_matrix(y_test, p3))
```

```python
#Count non zero coefficient
print("\nNumber of zero coefficient")
print("MLE:",np.sum(m1.coef_[0]==0))
print("MAP L2:", np.sum(m2.coef_[0] == 0))
print("MAP L1:", np.sum(m3.coef_[0] == 0))
```

```python
print("\n----Model Comaparison----")
models=["MLE","MAP L2","MAP L1"]
predictions=[p1,p2,p3]
for name,pred in zip(models,predictions):
    print("\n", name)
    print("Accuracy:", accuracy_score(y_test, pred))
    print("Precision:", precision_score(y_test, pred))
    print("Recall:", recall_score(y_test, pred))
    print("F1 Score:", f1_score(y_test, pred))
```

```python
comparison = pd.DataFrame({
    "Model": ["MLE", "MAP L2", "MAP L1"],
    "Accuracy": [
        accuracy_score(y_test, p1),
        accuracy_score(y_test, p2),
        accuracy_score(y_test, p3)
    ],
    "Precision": [
        precision_score(y_test, p1),
        precision_score(y_test, p2),
        precision_score(y_test, p3)
    ],
    "Recall": [
        recall_score(y_test, p1),
        recall_score(y_test, p2),
        recall_score(y_test, p3)
    ],
    "F1 Score": [
        f1_score(y_test, p1),
        f1_score(y_test, p2),
        f1_score(y_test, p3)
    ]
})
print(comparison.round(4))
```
