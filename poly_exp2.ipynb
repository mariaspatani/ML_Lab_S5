```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error,r2_score
```

```python
a=pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/mpg.csv")
print(a.head())
print(a.info())
```

```python
print(a.isnull().sum())
a=a[["displacement","mpg"]].dropna()
```

```python
x=a[["displacement"]]
y=a["mpg"]
```

```python
#split
x_train,x_test,y_train,y_test=train_test_split(
    x,y,random_state=42,test_size=0.2
)
print("Training Data: ",len(x_train))
print("Testing data: ",len(x_test))
```

```python
m1=LinearRegression()
m1.fit(x_train,y_train)
p1=m1.predict(x_test)

mse=mean_squared_error(y_test,p1)
r2=r2_score(y_test,p1
            )
print("\nLinear Regression")
print("-"*50)
print("MSE:",mse)
print("R2_score:",r2)
```

```python
result=[]
for d in range(2,6):
    poly=PolynomialFeatures(degree=d)
    x_train1=poly.fit_transform(x_train)
    x_test1=poly.transform(x_test)

    m2=LinearRegression()
    m2.fit(x_train1,y_train)
    p2=m2.predict(x_test1)

    mse= mean_squared_error(y_test,p2)
    r2=r2_score(y_test,p2)

    print("\nPolynomial Regression",d)
    print("-"*50)
    print("MSE:",mse)
    print("R2_score:",r2)
```

```python
result.append(("Degree:",d,mse,r2))
```

```python
comparison=pd.DataFrame(
    result, 
    columns=["Model", "Degree", "MSE", "R2 Score"]
)
print(comparison.round(4))

print("\nModel Comparison")
print(comparison.to_string(index=False))

best_mse=comparison.loc[comparison["MSE"].idxmin()]
best_r2=comparison.loc[comparison["R2 Score"].idxmax()]

print("\nBest Model based on MSE:", best_mse["Model"])
print("Lowest MSE:", best_mse["MSE"])

print("\nBest Model based on R2:", best_r2["Model"])
print("Highest R2:", best_r2["R2 Score"])
```
