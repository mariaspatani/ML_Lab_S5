```python

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error, r2_score

# 1. Load dataset
a = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/mpg.csv")

# 2. Remove missing values
a = a[["displacement", "mpg"]].dropna()

x = a[["displacement"]]
y = a["mpg"]

# 3. Split data
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)

results = []

# 4. Linear Regression
m1 = LinearRegression()
m1.fit(x_train, y_train)
p1 = m1.predict(x_test)

mse = mean_squared_error(y_test, p1)
r2 = r2_score(y_test, p1)

print("Linear Regression")
print("MSE:", mse)
print("R2:", r2)

results.append(["Linear", mse, r2])

# 5. Polynomial Regression - Degrees 2 to 5
for d in range(2, 6):
    poly = PolynomialFeatures(degree=d)

    x_train1 = poly.fit_transform(x_train)
    x_test1 = poly.transform(x_test)

    model = LinearRegression()
    model.fit(x_train1, y_train)
    pred = model.predict(x_test1)

    mse = mean_squared_error(y_test, pred)
    r2 = r2_score(y_test, pred)

    print("\nPolynomial Regression Degree", d)
    print("MSE:", mse)
    print("R2:", r2)

    results.append([f"Degree {d}", mse, r2])

# 6. Comparison Table
comparison = pd.DataFrame(
    results, columns=["Model", "MSE", "R2 Score"]
)

print("\nModel Comparison")
print(comparison.to_string(index=False))

# 7. Find best model
best_mse = comparison.loc[comparison["MSE"].idxmin()]
best_r2 = comparison.loc[comparison["R2 Score"].idxmax()]

print("\nBest Model based on MSE:", best_mse["Model"])
print("Lowest MSE:", best_mse["MSE"])

print("\nBest Model based on R2:", best_r2["Model"])
print("Highest R2:", best_r2["R2 Score"])
```
